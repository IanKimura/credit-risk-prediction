The dataset contains 307,511 loan applications, of which 24,825 (8.07%) have
`TARGET = 1`, indicating that the applicant experienced the repayment difficulty
defined by Home Credit. The remaining 91.93% have `TARGET = 0`.

The target variable is therefore imbalanced. A classifier that predicted
`TARGET = 0` for every applicant would achieve approximately 91.93% accuracy
while providing no useful ability to distinguish higher-risk applicants.

For this reason, accuracy will not be used as the primary evaluation metric.
Model evaluation will instead focus on discrimination and probability quality
using metrics such as ROC-AUC, PR-AUC, Brier score, calibration, and the
Kolmogorov-Smirnov (KS) statistic.

```python
missing_test_cols = [
    "EXT_SOURCE_1",
    "EXT_SOURCE_2",
    "EXT_SOURCE_3",
    "OCCUPATION_TYPE",
    "OWN_CAR_AGE",
    "AMT_REQ_CREDIT_BUREAU_YEAR",
    "COMMONAREA_AVG"
]

missing_target_results = []

for col in missing_test_cols:
    missing_mask = df[col].isna()

    missing_target_results.append({
        "Feature": col,
        "Missing_%": missing_mask.mean() * 100,
        "Target_Rate_When_Missing_%": df.loc[missing_mask, "TARGET"].mean() * 100,
        "Target_Rate_When_Observed_%": df.loc[~missing_mask, "TARGET"].mean() * 100
    })

missing_target_df = pd.DataFrame(missing_target_results)

missing_target_df
```

### Missingness and the Target

Missing values appear to contain potentially useful predictive information.
Several features show meaningful differences in observed TARGET rates depending
on whether the feature is missing.

For example, applicants missing `AMT_REQ_CREDIT_BUREAU_YEAR` have an observed
TARGET rate of approximately 10.34%, compared with 7.72% among applicants for
whom the feature is available. Missing `EXT_SOURCE_3` values are also associated
with a higher observed TARGET rate (9.31% vs. 7.77%).

However, the relationship is not uniform. Applicants with missing
`OCCUPATION_TYPE`, for example, have a lower observed TARGET rate than those
with an observed occupation (6.51% vs. 8.79%).

These results suggest that missingness may itself contain predictive information.
Imputation strategies should therefore consider preserving missingness indicators
rather than treating missing values solely as a data-cleaning problem.

### Age and Repayment Difficulty

Observed repayment difficulty decreases consistently with applicant age. The
TARGET rate falls from 11.44% among applicants aged 20–29 to 4.92% among
applicants aged 60–69.

The youngest age group therefore has an observed TARGET rate approximately
2.3 times that of the oldest age group. The monotonic pattern suggests that age
contains useful predictive information for credit-risk modeling.

This relationship is descriptive and does not imply that age itself causes
repayment difficulty.

# Loan characteristics

*   AMT_CREDIT = amount of credit for the loan
*   AMT_ANNUITY = the loan annuity/payment amount recorded for the application





```python
loan_cols = ["AMT_CREDIT", "AMT_ANNUITY"]

eda[loan_cols].describe(
    percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
).T
```

```python
for col in loan_cols:

    upper = eda[col].quantile(0.99)

    plt.figure(figsize=(8, 4))

    plt.hist(
        eda.loc[eda[col] <= upper, col].dropna(),
        bins=40
    )

    plt.xlabel(col)
    plt.ylabel("Number of Applicants")
    plt.title(f"Distribution of {col} (≤ 99th Percentile)")

    plt.show()
```

```python
for col in loan_cols:

    group_col = f"{col}_QUINTILE"

    eda[group_col] = pd.qcut(
        eda[col],
        q=5,
        labels=[
            "Q1 (Lowest)",
            "Q2",
            "Q3",
            "Q4",
            "Q5 (Highest)"
        ]
    )

    risk_table = (
        eda.groupby(group_col, observed=False)["TARGET"]
           .agg(["count", "mean"])
           .reset_index()
    )

    risk_table["Target_Rate_%"] = risk_table["mean"] * 100

    print(f"\n{col}")
    display(
        risk_table[
            [group_col, "count", "Target_Rate_%"]
        ]
    )
```

### Loan Amounts and Repayment Difficulty

Both credit amount and annuity exhibit non-linear relationships with repayment
difficulty. TARGET rates generally increase across the lower and middle
quintiles, but decline among applicants with the highest credit amounts and
annuities.

For credit amount, the observed TARGET rate peaks at approximately 10.1% in
the middle quintile before declining to 6.1% in the highest quintile.
Similarly, annuity TARGET rates rise through the fourth quintile before
falling to 6.7% among applicants with the highest annuity amounts.

These patterns suggest that absolute loan size and payment amount do not fully
capture an applicant's financial burden. Their relationship with repayment
difficulty may depend on the applicant's income and overall financial capacity.

# Financial Burden Ratios

```python
eda["CREDIT_INCOME_RATIO"] = (
    eda["AMT_CREDIT"] / eda["AMT_INCOME_TOTAL"]
)

eda["ANNUITY_INCOME_RATIO"] = (
    eda["AMT_ANNUITY"] / eda["AMT_INCOME_TOTAL"]
)

eda["CREDIT_ANNUITY_RATIO"] = (
    eda["AMT_CREDIT"] / eda["AMT_ANNUITY"]
)
```

Credit to Income: Credit Amount / Annual Income
- This measures the size of the loan relative to the applicant's income

Annuity to Income: Annuity / Annual Income
- This measures the payment burden relative to income

Credit to Annuity: Credit Amount / Annuity
- This gives us a rough scale of the loan relative to its scheduled payment


```python
ratio_cols = [
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "CREDIT_ANNUITY_RATIO"
]

eda[ratio_cols].describe(
    percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
).T
```

```python
for col in ratio_cols:

    group_col = f"{col}_QUINTILE"

    eda[group_col] = pd.qcut(
        eda[col],
        q=5,
        labels=[
            "Q1 (Lowest)",
            "Q2",
            "Q3",
            "Q4",
            "Q5 (Highest)"
        ],
        duplicates="drop"
    )

    risk_table = (
        eda.groupby(group_col, observed=False)["TARGET"]
           .agg(["count", "mean"])
           .reset_index()
    )

    risk_table["Target_Rate_%"] = risk_table["mean"] * 100

    print(f"\n{col}")

    display(
        risk_table[
            [group_col, "count", "Target_Rate_%"]
        ]
    )
```

### Financial Burden Ratios

Financial burden ratios provide additional information beyond the raw income,
credit, and annuity variables, although their relationships with repayment
difficulty are not uniformly monotonic.

The annuity-to-income ratio shows the clearest relationship with repayment
difficulty. The observed TARGET rate increases from 7.20% in the lowest
quintile to 8.70% in the fourth quintile, suggesting that applicants with
larger scheduled payments relative to their income generally experience
greater repayment difficulty.

Credit-to-income does not exhibit a monotonic relationship with TARGET.
Repayment difficulty increases through the middle quintiles before declining
among applicants with the highest credit-to-income ratios.

Credit-to-annuity also displays a nonlinear relationship. TARGET rates are
highest in the second and third quintiles and substantially lower in the
fourth and fifth quintiles.

These findings suggest that relative financial burden contains useful risk
information, but several relationships are nonlinear and may therefore be
better captured by flexible modeling approaches rather than assuming a
simple linear effect.

## 4. EDA Summary and Modeling Implications

The exploratory analysis identified several characteristics of the Home Credit
application data that will influence the modeling strategy.

### Target Distribution
The target variable is imbalanced, with approximately 8.1% of applicants
experiencing repayment difficulty. Because of this imbalance, model performance
should not be evaluated using accuracy alone. Metrics such as ROC-AUC, PR-AUC,
precision, recall, and probability calibration will be more informative.

### Data Quality and Missingness
Missing values are common throughout the dataset, with several variables
containing more than 50% missing observations. Missingness also appears to
contain information about repayment outcomes for some variables. Therefore,
high-missingness features should not automatically be discarded, and
missing-value handling will be incorporated into the preprocessing pipeline.

The DAYS_EMPLOYED variable contains the sentinel value 365243 for approximately
18% of applicants. This value does not represent a valid employment duration
and should be replaced with a missing value while preserving an indicator for
the anomaly.

### Applicant Characteristics
Applicant age and employment duration both show negative associations with
repayment difficulty. Younger applicants and applicants with shorter employment
histories generally exhibit higher observed TARGET rates.

Several categorical characteristics, including contract type, income type,
education, family status, and housing type, also show differences in observed
TARGET rates. Categories with very small sample sizes should be interpreted
cautiously.

### Financial Characteristics
Income alone shows relatively little linear relationship with TARGET, while
credit and annuity amounts display nonlinear relationships with repayment
difficulty.

Engineered financial burden measures, including credit-to-income,
annuity-to-income, and credit-to-annuity ratios, reveal additional patterns not
captured by the raw financial variables and will be retained as candidate
features.

### External Risk Scores
EXT_SOURCE_1, EXT_SOURCE_2, and EXT_SOURCE_3 exhibit some of the strongest
relationships with TARGET observed during the exploratory analysis. Higher
external score values are consistently associated with lower repayment
difficulty.

Missingness differs substantially across these variables, particularly for
EXT_SOURCE_1, so missing observations will need to be handled without simply
discarding affected applicants.

### Correlation and Redundancy
Several financial variables contain overlapping information. AMT_CREDIT and
AMT_ANNUITY are strongly correlated, while the engineered financial ratios are
also correlated with their underlying components.

These relationships do not necessarily require features to be removed, but
they should be considered when interpreting linear models such as logistic
regression.

### Modeling Implications
Based on the EDA, the modeling pipeline will:

1. Preserve the original raw dataset and construct a separate modeling dataset.
2. Correct known anomalous values while retaining useful anomaly indicators.
3. Handle numeric and categorical missing values systematically.
4. Engineer interpretable applicant and financial-risk features.
5. Encode categorical variables and transform numeric variables where appropriate.
6. Use stratified train/validation/test splits to preserve the target distribution.
7. Establish an interpretable logistic-regression baseline before evaluating
   more flexible machine-learning models.
8. Evaluate models using metrics appropriate for an imbalanced credit-risk problem.
